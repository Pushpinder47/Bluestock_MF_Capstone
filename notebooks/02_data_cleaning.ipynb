# Data Cleaning

## Objective
Clean the mutual fund dataset to improve data quality before analysis.

### Tasks Performed
- Removed duplicate records.
- Filled missing values.
- Converted date columns into datetime format.
- Saved the cleaned dataset into the processed folder.

### Expected Output
A cleaned dataset with no unnecessary missing values, ready for exploratory data analysis.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/raw/01_fund_master.csv")

print("Original Shape:", df.shape)

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing values safely
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].median())
    elif pd.api.types.is_datetime64_any_dtype(df[col]):
        continue
    else:
        df[col] = df[col].fillna("Unknown")
        
# Convert launch_date to datetime
df["launch_date"] = pd.to_datetime(df["launch_date"], errors="coerce")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nCleaned Shape:", df.shape)

# Save cleaned data
df.to_csv("../data/processed/clean_fund_master.csv", index=False)

print("\n✅ Cleaned file saved successfully.")
```
